```python
from datasets import load_dataset
from collections import Counter

ds = load_dataset("LabHC/bias_in_bios")
print(ds)  # train / dev / test, kolommen: hard_text, profession, gender

professions = [
    "accountant", "architect", "attorney", "chiropractor", "comedian", "composer",
    "dentist", "dietitian", "dj", "filmmaker", "interior_designer", "journalist",
    "model", "nurse", "painter", "paralegal", "pastor", "personal_trainer",
    "photographer", "physician", "poet", "professor", "psychologist", "rapper",
    "software_engineer", "surgeon", "teacher", "yoga_teacher",
]
genders = ["male", "female"]

train = ds["train"]

# 1. Lege teksten?
for split in ds:
    print(split, "lege teksten:", sum(1 for t in ds[split]["hard_text"] if not t or not t.strip()))

# 2. Verdeling van beroepen (train)
prof_counts = Counter(train["profession"])
for idx, count in prof_counts.most_common():
    print(f"{professions[idx]:18s} {count:7d}  ({count/len(train):.1%})")

# 3. Verdeling van gender
gender_counts = Counter(train["gender"])
print({genders[k]: f"{v} ({v/len(train):.1%})" for k, v in gender_counts.items()})

# 4. Aandeel vrouwen per beroep (het belangrijkste voor de bias-analyse)
female_per_prof = Counter(p for p, g in zip(train["profession"], train["gender"]) if g == 1)
print("\nAandeel vrouwen per beroep:")
for idx in sorted(prof_counts, key=lambda i: female_per_prof[i] / prof_counts[i]):
    print(f"{professions[idx]:18s} {female_per_prof[idx]/prof_counts[idx]:.1%}")

# 5. Tekstlengte in woorden
lengths = [len(t.split()) for t in train["hard_text"]]
print(f"\nGemiddeld {sum(lengths)/len(lengths):.1f} woorden, min {min(lengths)}, max {max(lengths)}")

# 6. Verdeling per split vergelijken (zijn de splits vergelijkbaar?)
for split in ds:
    c = Counter(ds[split]["profession"])
    top3 = [(professions[i], f"{n/len(ds[split]):.1%}") for i, n in c.most_common(3)]
    print(split, top3)

# 7. Voorbeelden
for ex in train.shuffle(seed=42).select(range(5)):
    print(f"[{professions[ex['profession']]}, {genders[ex['gender']]}] {ex['hard_text'][:200]}")
```

```python
## token indicators
import numpy as np
from tokenizers import Tokenizer
from tokenizers.pre_tokenizers import Metaspace
from tokenizers.trainers import BpeTrainer
from tokenizers.models import BPE

def train_tokenizer(corpus, vocab_size, min_frequency, pretokenizer=Metaspace(), unk_token="[UNK]"):
    """
    corpus: a list of texts (each a string)
    vocab_size: number of distinct BPE tokens
    min_frequency: we only merge token pairs that are frequent enough
    pretokenizer: a simple tokenizer that we use before BPE tokenization
    
        You probably don't need to change the pretokenizer, unless you are working with
        languages that really don't use any form of blank space 
        (then you could, for example, pretokenize at character level)
    """
    # Create an empty BPE tokenizer
    tokenizer = Tokenizer(BPE(unk_token=unk_token)) #unk_token used for characters that cannot be encoded

    # First split the corpus into words
    tokenizer.pre_tokenizer = pretokenizer

    # Tell BPE how large the vocabulary should be
    trainer = BpeTrainer(
        vocab_size=vocab_size, # The size of the vocabulary to learn.
        min_frequency=min_frequency, # Only perform a merge if the pair occurs at least min_frequency times in the training corpus.
        special_tokens=[unk_token]
    )

    # Learn the vocabulary and merge rules
    tokenizer.train_from_iterator(corpus, trainer=trainer)

    return tokenizer

class Encoder:

    def __len__(self) -> int:
        """Dimensionality of the representation space"""
        raise NotImplementedError("Concrete implementations will be discussed later")

    def __call__(self, x: str) -> 'np.array':
        """Represent x as a D-dimensional vector of real numbers"""
        raise NotImplementedError("Concrete implementations will be discussed later")
    
class TokenIndicator(Encoder):

    def __init__(self, tokenizer: Tokenizer):
        self.tokenizer = tokenizer
        
    def __len__(self) -> int:
        """Dimensionality of the representation space"""
        return len(self.tokenizer.get_vocab())

    def __call__(self, x: str) -> 'np.array':
        """Represent x as a D-dimensional vector of real numbers"""
        unique_token_ids = list(set(self.tokenizer.encode(x).ids))

        # A 0 vector with the same length as the vocabulary except in positions corresponding 
        # to tokens appearing in the document
        vector = np.zeros(len(self))
        vector[unique_token_ids] = 1

        return vector
```

```python
vocab_size = 1000
min_frequency = 2

corpus = ds["train"]["hard_text"]

print(f"Training on {len(corpus)} samples...")
tokenizer = train_tokenizer(corpus, vocab_size, min_frequency)
print("Tokennizer Trained")

vocab = tokenizer.get_vocab()
token_indicator_encoder = TokenIndicator(tokenizer)
doc_repr = token_indicator_encoder(corpus[0])
print(len(doc_repr))
print(doc_repr.sum())
```
